```python
# tiny_optimizer.py
import pulp

# 1. THE SITUATION (the facts we know)
solar = [0,15,15,0]
price = [20, 20, 80, 80]   # $/MWh for 4 hours: cheap, cheap, expensive, expensive
demand = 10                # city needs 10 MW every hour
cap = 20                   # battery can hold 20 MWh
rate = 10 # battery can charge/discharge max 10 MW
line = 15
sell_factor = 0.9
T = range(4)               # hours 0, 1, 2, 3

# 2. CREATE THE PROBLEM ("I want to minimize something")
m = pulp.LpProblem("tiny", pulp.LpMinimize)

# 3. VARIABLES (the things the optimizer is allowed to decide)
buy       = pulp.LpVariable.dicts("buy", T, line)                 # MW bought from grid
sell      = pulp.LpVariable.dicts("sell",T,0,line)
charge    = pulp.LpVariable.dicts("charge", T, lowBound=0, upBound=rate)
discharge = pulp.LpVariable.dicts("discharge", T, lowBound=0, upBound=rate)
soc       = pulp.LpVariable.dicts("soc", range(5), lowBound=0, upBound=cap)  # battery level
s_used = pulp.LpVariable.dicts("s_used",T,0)
# 4. CONSTRAINTS (rules that must be true)
m += soc[0] == 0                                   # battery starts empty
for t in T:
    m += s_used[t] <= solar[t]# power in = power out
    m += s_used[t] + buy[t] + discharge[t] == demand + charge[t] + sell[t]
    m += soc[t+1] == soc[t]+ charge[t] - discharge[t]# battery level updates

# 5. THE GOAL (what to minimize: total money spent)
m += pulp.lpSum(buy[t] * price[t] - sell[t]*price[t] * sell_factor for t in T)

# 6. SOLVE AND PRINT
m.solve(pulp.PULP_CBC_CMD(msg=0))
print(pulp.LpStatus[m.status])
for t in T:
    print(f"hour {t}: price={price[t]}  buy={buy[t].value()}  "
          f"charge={charge[t].value()}  discharge={discharge[t].value()}  "
          f"battery_after={soc[t+1].value()}")
print("total cost:", pulp.value(m.objective))
```
